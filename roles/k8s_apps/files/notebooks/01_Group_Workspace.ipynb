# Group workspace

This notebook confirms which persistent group workspace you are using.
Your group files are isolated from the other group's workspace.

```python
import os
from pathlib import Path

username = os.environ.get('JUPYTERHUB_USER', '<unknown>')
groups = {'group_01': 'Group 01', 'group_02': 'Group 02'}
group = groups.get(username.lower(), 'Unassigned test workspace')
workspace = Path('/home/jovyan/digitafrica')

print(f'JupyterHub login: {username}')
print(f'Workshop group: {group}')
print(f'Workspace: {workspace}')
print('This workspace is persistent for this JupyterHub username.')
```
